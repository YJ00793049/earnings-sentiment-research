```python
# === Script 1: Get S&P 500 companies + their 8-K filings from EDGAR ===
import requests, pandas as pd, time
from io import StringIO

HEADERS = {"User-Agent": "Yuvraj Jindal research yuvrajjindal2020@gmail.com"}
WIKI_HEADERS = {"User-Agent": "Mozilla/5.0 (research project)"}

# 1) Get current S&P 500 tickers from Wikipedia
wiki_url = "https://en.wikipedia.org/wiki/List_of_S%26P_500_companies"
html = requests.get(wiki_url, headers=WIKI_HEADERS).text
sp500 = pd.read_html(StringIO(html))[0]
sp500["Symbol"] = sp500["Symbol"].str.replace(".", "-", regex=False)
tickers = sp500["Symbol"].tolist()
print(f"Loaded {len(tickers)} S&P 500 tickers")

# 2) Map tickers -> CIK
cik_map = requests.get("https://www.sec.gov/files/company_tickers.json", headers=HEADERS).json()
ticker_to_cik = {v["ticker"]: str(v["cik_str"]).zfill(10) for v in cik_map.values()}

# 3) Pull recent filings, keep the 8-Ks from 2022-2024
rows = []
for tk in tickers:
    cik = ticker_to_cik.get(tk)
    if not cik:
        continue
    try:
        url = "https://data.sec.gov/submissions/CIK" + cik + ".json"
        data = requests.get(url, headers=HEADERS).json()
        recent = data["filings"]["recent"]
        for form, date, accno, doc in zip(
            recent["form"], recent["filingDate"],
            recent["accessionNumber"], recent["primaryDocument"]
        ):
            if form == "8-K" and "2022-01-01" <= date <= "2024-12-31":
                rows.append({
                    "ticker": tk, "cik": cik, "form": form,
                    "filing_date": date, "accession": accno, "primary_doc": doc
                })
    except Exception as e:
        print("skip " + tk + ": " + str(e))
    time.sleep(0.12)

df = pd.DataFrame(rows)
df.to_csv("sp500_8k_filings.csv", index=False)
print(f"Saved {len(df)} 8-K filings to sp500_8k_filings.csv")
df.head()
```

```python
# === Script 2: Find the earnings releases (Item 2.02) among the 8-Ks ===
import requests, pandas as pd, time

HEADERS = {"User-Agent": "Yuvraj Jindal research yuvrajjindal2020@gmail.com"}

# EDGAR tags every 8-K with item codes. Item 2.02 = "Results of Operations
# and Financial Condition" = the earnings release. That's the one we want.
# We re-pull the submissions feed because we now also need the "items" tag
# and the document name so we can build a link to each press release.

sp500 = pd.read_csv("sp500_8k_filings.csv")
companies = sp500[["ticker", "cik"]].drop_duplicates()

rows = []
for _, r in companies.iterrows():
    tk = r["ticker"]
    cik = str(r["cik"]).zfill(10)
    try:
        url = "https://data.sec.gov/submissions/CIK" + cik + ".json"
        data = requests.get(url, headers=HEADERS).json()
        recent = data["filings"]["recent"]
        for form, accno, date, items, doc in zip(
            recent["form"], recent["accessionNumber"], recent["filingDate"],
            recent["items"], recent["primaryDocument"]
        ):
            if form == "8-K" and "2022-01-01" <= date <= "2024-12-31" and "2.02" in items:
                acc_nodash = accno.replace("-", "")
                doc_url = ("https://www.sec.gov/Archives/edgar/data/"
                           + str(int(cik)) + "/" + acc_nodash + "/" + doc)
                rows.append({
                    "ticker": tk, "cik": cik, "filing_date": date,
                    "accession": accno, "primary_doc": doc, "doc_url": doc_url
                })
    except Exception as e:
        print("skip", tk, "->", str(e))
    time.sleep(0.12)

earnings = pd.DataFrame(rows)
earnings.to_csv("earnings_8k.csv", index=False)
print("Found", len(earnings), "earnings releases (Item 2.02)")
earnings.head()
```

```python
# === Script 3: Sample earnings releases and download the press-release text ===
import requests, pandas as pd, time
from bs4 import BeautifulSoup

HEADERS = {"User-Agent": "Yuvraj Jindal research yuvrajjindal2020@gmail.com"}

earnings = pd.read_csv("earnings_8k.csv")

# Sample a clean, manageable set. random_state makes it 100% reproducible.
SAMPLE_SIZE = 250
sample = earnings.sample(n=SAMPLE_SIZE, random_state=42).reset_index(drop=True)

def get_press_release_text(cik, accession, fallback_doc):
    cik_int = str(int(cik))
    acc_nodash = accession.replace("-", "")
    base = "https://www.sec.gov/Archives/edgar/data/" + cik_int + "/" + acc_nodash + "/"
    target = fallback_doc
    try:
        listing = requests.get(base + "index.json", headers=HEADERS).json()
        items = listing["directory"]["item"]
        ex = [it["name"] for it in items
              if "99" in it["name"].lower()
              and "index" not in it["name"].lower()
              and it["name"].lower().endswith((".htm", ".html", ".txt"))]
        if ex:
            target = ex[0]
    except Exception:
        pass
    html = requests.get(base + target, headers=HEADERS).text
    return BeautifulSoup(html, "html.parser").get_text(separator=" ", strip=True)

texts = []
for i, r in sample.iterrows():
    try:
        texts.append(get_press_release_text(r["cik"], r["accession"], r["primary_doc"]))
    except Exception as e:
        print("skip", r["ticker"], r["filing_date"], "->", str(e))
        texts.append("")
    time.sleep(0.2)

sample["text"] = texts
sample = sample[sample["text"].str.len() > 200].reset_index(drop=True)
sample.to_csv("earnings_text.csv", index=False)
print("Downloaded text for", len(sample), "earnings releases")
print("Median length:", int(sample["text"].str.len().median()), "characters")
sample[["ticker", "filing_date"]].head()
```

```python
# === Script 4: FinBERT sentiment scoring of each earnings press release ===
!pip install -q transformers

import pandas as pd, numpy as np, torch, re
from transformers import AutoTokenizer, AutoModelForSequenceClassification

df = pd.read_csv("earnings_text.csv")

tok = AutoTokenizer.from_pretrained("ProsusAI/finbert")
model = AutoModelForSequenceClassification.from_pretrained("ProsusAI/finbert")
model.eval()
use_gpu = torch.cuda.is_available()
if use_gpu:
    model = model.cuda()
print("Using GPU:", use_gpu)

id2label = model.config.id2label
pos_idx = [i for i, l in id2label.items() if l.lower() == "positive"][0]
neg_idx = [i for i, l in id2label.items() if l.lower() == "negative"][0]

def split_sentences(text):
    sents = re.split(r"(?<=[.!?])\s+", str(text))
    return [s.strip() for s in sents if len(s.strip()) > 20]

def score_text(text, max_sents=60, batch=16):
    sents = split_sentences(text)[:max_sents]
    if not sents:
        return np.nan
    diffs = []
    for k in range(0, len(sents), batch):
        chunk = sents[k:k+batch]
        enc = tok(chunk, return_tensors="pt", padding=True, truncation=True, max_length=128)
        if use_gpu:
            enc = {kk: vv.cuda() for kk, vv in enc.items()}
        with torch.no_grad():
            logits = model(**enc).logits
        probs = torch.softmax(logits, dim=1).cpu().numpy()
        diffs.extend(list(probs[:, pos_idx] - probs[:, neg_idx]))
    return float(np.mean(diffs))

scores = []
for i, r in df.iterrows():
    scores.append(score_text(r["text"]))

df["finbert_sentiment"] = scores
df_out = df.drop(columns=["text"])
df_out.to_csv("earnings_sentiment.csv", index=False)
print("Scored", df["finbert_sentiment"].notna().sum(), "earnings releases")
print("Mean sentiment:", round(df["finbert_sentiment"].mean(), 4))
print("Range:", round(df["finbert_sentiment"].min(), 3), "to", round(df["finbert_sentiment"].max(), 3))
df_out[["ticker", "filing_date", "finbert_sentiment"]].head(10)
```

```python
# === Script 5: Stock price reaction (abnormal returns) around each earnings release ===
!pip install -q yfinance

import pandas as pd, numpy as np, yfinance as yf

df = pd.read_csv("earnings_sentiment.csv")
df["filing_date"] = pd.to_datetime(df["filing_date"])
tickers = list(df["ticker"].unique())
print("Events:", len(df), "| Unique tickers:", len(tickers))

# One bulk download of all tickers + the S&P 500 benchmark
px = yf.download(tickers + ["SPY"], start="2021-12-01", end="2025-01-15",
                 progress=False, auto_adjust=True)["Close"]
rets = px.pct_change()
spy_ret = rets["SPY"]
print("Downloaded prices for", rets.shape[1], "symbols")

def car(ar, event_date, lo, hi):
    idx = ar.index
    pos = idx.searchsorted(event_date)
    start, end = pos + lo, pos + hi
    if pos >= len(idx) or start < 0 or end >= len(idx):
        return np.nan
    return float(ar.iloc[start:end+1].sum())

rows = []
for _, r in df.iterrows():
    tk = r["ticker"]
    if tk not in rets.columns:
        continue
    ar = (rets[tk] - spy_ret).dropna()   # market-adjusted abnormal return
    ed = r["filing_date"]
    rows.append({
        "ticker": tk,
        "filing_date": ed,
        "finbert_sentiment": r["finbert_sentiment"],
        "car_m1_p1": car(ar, ed, -1, 1),
        "car_0_p1":  car(ar, ed, 0, 1),
        "car_0_p5":  car(ar, ed, 0, 5),
    })

res = pd.DataFrame(rows).dropna().reset_index(drop=True)
res.to_csv("event_returns.csv", index=False)
try:
    res.to_csv("/content/drive/MyDrive/earnings_research/event_returns.csv", index=False)
except Exception:
    pass
print("Events with returns:", len(res))
print(res[["car_m1_p1", "car_0_p1", "car_0_p5"]].describe().round(4))
res.head()
```

```python
# === Script 6: Does FinBERT sentiment predict the market reaction? ===
import pandas as pd, numpy as np
import statsmodels.api as sm
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, spearmanr

res = pd.read_csv("event_returns.csv")
windows = ["car_m1_p1", "car_0_p1", "car_0_p5"]
labels = {"car_m1_p1": "[-1,+1]", "car_0_p1": "[0,+1]", "car_0_p5": "[0,+5]"}

print("=== Correlations: sentiment vs abnormal return ===")
for w in windows:
    pr, pp = pearsonr(res["finbert_sentiment"], res[w])
    sr, sp = spearmanr(res["finbert_sentiment"], res[w])
    print(labels[w], "Pearson r =", round(pr, 3), "(p =", round(pp, 4),
          ") | Spearman =", round(sr, 3), "(p =", round(sp, 4), ")")

print("\n=== Regression: abnormal return ~ sentiment (Newey-West SE) ===")
X = sm.add_constant(res["finbert_sentiment"])
for w in windows:
    model = sm.OLS(res[w], X).fit(cov_type="HAC", cov_kwds={"maxlags": 5})
    beta = model.params["finbert_sentiment"]
    t = model.tvalues["finbert_sentiment"]
    p = model.pvalues["finbert_sentiment"]
    print(labels[w], ": beta =", round(beta, 4), "| t =", round(t, 2),
          "| p =", round(p, 4), "| R^2 =", round(model.rsquared, 4))

# Scatter + fit line for the main window
w = "car_m1_p1"
b, a = np.polyfit(res["finbert_sentiment"], res[w], 1)
xs = np.linspace(res["finbert_sentiment"].min(), res["finbert_sentiment"].max(), 100)
plt.figure(figsize=(8, 6))
plt.scatter(res["finbert_sentiment"], res[w], alpha=0.5, s=25)
plt.plot(xs, b * xs + a, color="red", lw=2, label=f"fit slope = {b:.3f}")
plt.axhline(0, color="gray", lw=0.8)
plt.axvline(0, color="gray", lw=0.8)
plt.xlabel("FinBERT sentiment (press-release tone)")
plt.ylabel("Cumulative abnormal return [-1,+1]")
plt.title("Earnings press-release sentiment vs. stock reaction")
plt.legend()
plt.tight_layout()
plt.savefig("sentiment_vs_return.png", dpi=150)
try:
    plt.savefig("/content/drive/MyDrive/earnings_research/sentiment_vs_return.png", dpi=150)
except Exception:
    pass
plt.show()
print("Saved chart: sentiment_vs_return.png")
```

```python
# === Script 7: Add the earnings-surprise (SUE) control ===
import pandas as pd, numpy as np, yfinance as yf
import statsmodels.api as sm

res = pd.read_csv("event_returns.csv")
res["filing_date"] = pd.to_datetime(res["filing_date"])

# Pull earnings surprise (%) for each ticker from Yahoo (free)
surprises = {}
for tk in res["ticker"].unique():
    try:
        ed = yf.Ticker(tk).get_earnings_dates(limit=40)
        if ed is not None and "Surprise(%)" in ed.columns:
            tmp = ed[["Surprise(%)"]].dropna().copy()
            tmp.index = pd.to_datetime(tmp.index).tz_localize(None)
            surprises[tk] = tmp
    except Exception:
        pass
print("Got earnings data for", len(surprises), "tickers")

def match_surprise(tk, date, tol_days=5):
    if tk not in surprises:
        return np.nan
    s = surprises[tk]
    deltas = np.abs((s.index - date).days.to_numpy())
    if len(deltas) == 0:
        return np.nan
    j = int(deltas.argmin())
    return float(s["Surprise(%)"].iloc[j]) if deltas[j] <= tol_days else np.nan

res["surprise_pct"] = [match_surprise(r["ticker"], r["filing_date"]) for _, r in res.iterrows()]
matched = res.dropna(subset=["surprise_pct"]).reset_index(drop=True)
print("Events matched with an earnings surprise:", len(matched), "of", len(res))

print("\n=== CAR[0,+1] ~ sentiment + earnings surprise (Newey-West SE) ===")
X = sm.add_constant(matched[["finbert_sentiment", "surprise_pct"]])
m = sm.OLS(matched["car_0_p1"], X).fit(cov_type="HAC", cov_kwds={"maxlags": 5})
print(m.summary())
matched.to_csv("event_returns_with_surprise.csv", index=False)
try:
    matched.to_csv("/content/drive/MyDrive/earnings_research/event_returns_with_surprise.csv", index=False)
except Exception:
    pass
```

```python
# === Script 8: Full results table (all windows, with control + winsorized) ===
import pandas as pd, numpy as np
import statsmodels.api as sm
from scipy.stats.mstats import winsorize

res = pd.read_csv("event_returns_with_surprise.csv")
windows = {"car_m1_p1": "[-1,+1]", "car_0_p1": "[0,+1]", "car_0_p5": "[0,+5]"}

# Winsorize CARs at 1%/99% to tame outliers (robustness check)
for w in windows:
    res[w + "_w"] = np.asarray(winsorize(res[w].to_numpy(), limits=[0.01, 0.01]))

def run(y, xcols):
    X = sm.add_constant(res[xcols])
    return sm.OLS(res[y], X).fit(cov_type="HAC", cov_kwds={"maxlags": 5})

print("CAR ~ sentiment (+surprise), Newey-West SE, winsorized\n")
print(f"{'Window':<9}{'Spec':<16}{'beta':>9}{'t':>7}{'p':>8}{'R2':>7}{'N':>6}")
for w, lab in windows.items():
    for spec, cols in [("sentiment", ["finbert_sentiment"]),
                       ("sent+surprise", ["finbert_sentiment", "surprise_pct"])]:
        m = run(w + "_w", cols)
        b = m.params["finbert_sentiment"]; t = m.tvalues["finbert_sentiment"]
        p = m.pvalues["finbert_sentiment"]; r2 = m.rsquared; n = int(m.nobs)
        print(f"{lab:<9}{spec:<16}{b:>9.4f}{t:>7.2f}{p:>8.4f}{r2:>7.3f}{n:>6}")
```
